### Import libraries

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

### Read and display original image

```python
img = cv2.imread('image.png')
plt.imshow(img[:, :, ::-1])
plt.title("Original Image")
plt.show()
```

### Aplly Gaussian Filter

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
gaussian = cv2.GaussianBlur(gray, (5, 5), 0)
plt.imshow(gaussian, cmap='gray')
plt.title("Gaussian Filter")
plt.axis('off')
plt.show()
```

### Sobel Edge Detector

```python
sobel = cv2.Sobel(gaussian, cv2.CV_64F, 1, 0, ksize=3)

plt.imshow(abs(sobel), cmap='gray')
plt.title("Sobel")
plt.show()
```

### Prewitt Edge Detector

```python
kernel = np.array([[-1, 0, 1],
                   [-1, 0, 1],
                   [-1, 0, 1]])

prewitt = cv2.filter2D(gaussian, -1, kernel)

plt.imshow(prewitt, cmap='gray')
plt.title("Prewitt")
plt.show()
```

### Roberts Edge Detector

```python
kernel = np.array([[1, 0],
                   [0, -1]])

roberts = cv2.filter2D(gaussian, -1, kernel)

plt.imshow(roberts, cmap='gray')
plt.title("Roberts")
plt.show()
```

### Laplacian Edge Detector

```python
laplacian = cv2.Laplacian(gaussian, cv2.CV_64F)

plt.imshow(abs(laplacian), cmap='gray')
plt.title("Laplacian")
plt.show()
```

### Canny Edge Detector

```python
canny = cv2.Canny(gaussian, 100, 200)

plt.imshow(canny, cmap='gray')
plt.title("Canny")
plt.show()
```

### Display all outputs together

```python
plt.figure(figsize=(14, 10))

plt.subplot(3, 3, 1)
plt.imshow(img[:, :, ::-1])
plt.title("Original Image")

plt.subplot(3, 3, 2)
plt.imshow(gaussian, cmap='gray')
plt.title("Gaussian Filter")

plt.subplot(3, 3, 3)
plt.imshow(abs(sobel), cmap='gray')
plt.title("Sobel")

plt.subplot(3, 3, 4)
plt.imshow(prewitt, cmap='gray')
plt.title("Prewitt")

plt.subplot(3, 3, 5)
plt.imshow(roberts, cmap='gray')
plt.title("Roberts")

plt.subplot(3, 3, 6)
plt.imshow(abs(laplacian), cmap='gray')
plt.title("Laplacian")

plt.subplot(3, 3, 7)
plt.imshow(canny, cmap='gray')
plt.title("Canny")

plt.tight_layout()
plt.show()
```
